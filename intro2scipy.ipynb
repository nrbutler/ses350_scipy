# SciPy Demonstration

This notebook demonstrates three common capabilities of the **SciPy** library:
1. **Integration** (using `scipy.integrate`)
2. **Optimization** (using `scipy.optimize`)
3. **Linear Algebra** (using `scipy.linalg`)

## Example 1: Numerical Integration
We will evaluate the definite integral:
$$\int_{0}^{\pi} \sin(x) \, dx$$

```python
import numpy as np
from scipy import integrate

# Define the function to integrate
f = lambda x: np.sin(x)

# Perform the integration from 0 to pi
result, error = integrate.quad(f, 0, np.pi)

print(f"Result of integration: {result}")
print(f"Estimated error: {error}")
```

## Example 2: Optimization
We will find the minimum of a simple scalar function:
$$f(x) = x^2 + 2x + 2$$

```python
from scipy import optimize

# Define the quadratic function
def f(x):
    return x**2 + 2*x + 2

# Find the minimum starting from an initial guess of x = 0
result = optimize.minimize(f, x0=[0])

print(f"Success: {result.success}")
print(f"Minimizer (x): {result.x[0]}")
print(f"Minimum value (f(x)): {result.fun}")
```

## Example 3: Linear Algebra
We will solve the system of linear equations $Ax = b$:
$$
\begin{pmatrix}
3 & 2 \\
1 & 2
\end{pmatrix}
\begin{pmatrix}
x_1 \\
x_2
\end{pmatrix}
=
\begin{pmatrix}
2 \\
4
\end{pmatrix}
$$

```python
from scipy import linalg

# Define matrix A and vector b
A = np.array([[3, 2],
              [1, 2]])
b = np.array([2, 4])

# Solve the linear system
x = linalg.solve(A, b)

print(f"Solution x: {x}")
# Verify the solution by matrix multiplication
print(f"Verification (A * x): {A.dot(x)}")
```
